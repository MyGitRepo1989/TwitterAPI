```python
def clean_tweet(tweet):
    if type(tweet) == np.float:
        return ""
    temp=str(tweet)
    #temp = tweet.lower()
    temp = re.sub("'", "", temp)
    temp = re.sub("RT","", temp)
    # to avoid removing contractions in english
    temp = re.sub("@[A-Za-z0-9_]+","", temp)
    temp = re.sub("#[A-Za-z0-9_]+","", temp)
    temp = re.sub(r'http\S+', '', temp)
    temp = re.sub('[()!?,]', '', temp)
    temp = re.sub('\[.*?\]','', temp)
    #temp = re.sub("[^a-z0-9]"," ", temp)
    temp = temp.split()
    #temp = [w for w in temp if not w in stopwords]
    temp = " ".join(word for word in temp)
    return temp
```

```python
from nltk.corpus import stopwords
stopwords_dict = {word: 1 for word in stopwords.words("english")}
```

```python
def cleantweets(tweets):
    topic_clean=[]
    for i in range(len(tweets)):
        #print(topic_text[i].text,i)

        s = tweets[i]
        s = " ".join([word for word in s.split() if word not in stopwords_dict])
        s =re.sub('RT', '', s)
        s =re.sub('https://\w+.\w+/\w+', '', s)
        s =re.sub('\@\w+','', s)

        s =re.sub('\#\w+', '', s)
        s=re.sub('\n', '', s)
        s =emoji.replace_emoji(s)
        s =re.sub('[^A-Za-z0-9]+', ' ', s)
        s =re.sub('\n+ ', '', s)
        topic_clean.append(s)
    return topic_clean
```

```python
classifier = pipeline("text-classification", model=model, tokenizer=tokenizer,return_all_scores=True)
def getsentiment(textdata):
    tweettext =[]
    cleantext=[]
    great_s=0
    neutral_s=0
    bad_s=0
    for tweet in textdata:
        tweettext.append(tweet.text)
    cleantext = cleantweets(tweettext)
    #print(len(cleantext))
    for tweet_s in cleantext:
        senti= pd.DataFrame(classifier(tweet_s)[0]).sort_values(by="score",ascending=False)[:3]['label'].values[0]
        #print(senti)

        if senti == "5 stars" :
            #print("TRUE")
            great_s= great_s +1
        if senti == "4 stars" :
            #print("TRUE")
            great_s= great_s +1
        if senti == "3 stars" :
            #print("TRUE")
            neutral_s= neutral_s +1
        if senti == "2 stars" :
            #print("TRUE")
            bad_s= bad_s +1
        if senti == "1 star" :
            #print("TRUE")
            bad_s= bad_s +1

    
    #print(great_s +neutral_s + bad_s)    
    return great_s,neutral_s,bad_s
    
    
```
